```python
import pandas as pd
df=pd.read_csv("placement_dataset.csv")
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))

plt.hist(df["CGPA"].dropna(),
         bins=15,
         color="skyblue",
         edgecolor="black")

plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")

plt.show()
```
