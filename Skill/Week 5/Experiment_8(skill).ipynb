```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("StudentsPerformance.csv")

# Select only two numerical features
X = df[['math score', 'reading score']]

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['math score', 'reading score'])

print("Original Data")
print(X.head())

print("\nAfter standard Scaling")
print(scaled_df.head())
```

```python
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

# Fill missing categorical values with mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0]) # Changed to reassign instead of inplace=True

df.to_csv("placementfilledcat.csv", index=False)
print(df.isnull().sum())
```
